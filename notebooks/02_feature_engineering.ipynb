```python
import pandas as pd
from src.data_loading.loader import load_data
from src.preprocessing.preprocessor import clean
from src.features.feature_engineering import build_features, feature_columns

cleaned = clean(load_data(config), config)
features = build_features(cleaned, config, save=False)
print(f'{len(feature_columns(features))} features')
features.head()
```

```python
# Leakage sanity check: lag_1 must equal the previous day's sales
g = features[features['series_id'] == features['series_id'].iloc[0]].sort_values('date')
assert (g['lag_1'].to_numpy()[1:] == g['sales'].to_numpy()[:-1]).all()
print('lag_1 is strictly past — no leakage')
```
